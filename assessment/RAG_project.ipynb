```python
# Data Loading: First Step

from langchain_community.document_loaders import PyPDFLoader

documents = [
    "mevita.pdf",
    "mevion.pdf",
    "mevius.pdf",
    "adaptive_learning.pdf",
    "VLA.pdf"
]

paperdata = []

for pdf_file in documents:
    loader = PyPDFLoader(pdf_file)
    paperdata.extend(loader.load())

print("Total pages loaded:", len(paperdata))
print(paperdata[0].page_content[:500])
```

```python
#Text Splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
textSplitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)
finalDocuments=textSplitter.split_documents(paperdata)
finalDocuments
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
print(embeddings)
```

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(finalDocuments, embeddings)
db
```

```python
query = input("Enter you question for the Research paper (Attention Is All You Need) : ")
docs = db.similarity_search(query)
for i in range(len(docs)):
    print(docs[i].page_content)
```
